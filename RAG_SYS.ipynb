```python
import re
from langchain_community.document_loaders import PyPDFLoader

# Load PDF
pdf_path = list(uploaded.keys())[0]
loader = PyPDFLoader(pdf_path)
documents = loader.load()

# Clean text
def clean_text(text):
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

for doc in documents:
    doc.page_content = clean_text(doc.page_content)

# Chunking
try:
    from langchain_text_splitters import RecursiveCharacterTextSplitter
except ImportError:
    from langchain.text_splitter import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print(f"Pages loaded: {len(documents)}")
print(f"Chunks created: {len(chunks)}")
print("\nSample chunk:")
print(chunks[0].page_content[:1000])
print("\nMetadata:")
print(chunks[0].metadata)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
sample_text = chunks[0].page_content
vector = embeddings.embed_query(sample_text)

print(type(vector))
print(len(vector))
print(vector[:10])
```

```python
chunk_vectors = embeddings.embed_documents([chunk.page_content for chunk in chunks])

print(f"Number of vectors: {len(chunk_vectors)}")
print(f"Vector dimension: {len(chunk_vectors[0])}")
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(chunks, embeddings)
print("FAISS index created successfully.")
```

```python
query = "What is this document about?"
results = vectorstore.similarity_search(query, k=3)

for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content[:500])
    print("Metadata:", doc.metadata)
```

```python
vectorstore.save_local("faiss_index")
print("FAISS index saved.")
```

```python
vectorstore = FAISS.load_local(
    "faiss_index",
    embeddings,
    allow_dangerous_deserialization=True
)

print("FAISS index loaded.")
```

```python
query = "What is the main topic?"
results = vectorstore.similarity_search(query, k=3)

print(len(results))
print(results[0].metadata)
print(results[0].page_content[:300])
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
print("Retriever created successfully.")
```

```python
query = "What topics are discussed in the document?"
docs = retriever.invoke(query)

print("Number of docs returned:", len(docs))
print(docs[0].metadata)
print(docs[0].page_content[:300])
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are a helpful assistant answering questions based only on the provided context from a PDF document.

Rules:
- Use only the given context.
- Do not invent or guess information.
- If the answer is not in the context, say: "I don't have enough information in the document."
- Answer clearly and concisely.
- If possible, mention the page numbers used as sources.

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
def format_docs(docs):
    return "\n\n".join(
        f"[Page {doc.metadata.get('page', 'N/A')}] {doc.page_content}"
        for doc in docs
    )
```

```python
query = "What topics are discussed in the document?"
docs = retriever.invoke(query)
context = format_docs(docs)

messages = prompt.format_messages(context=context, question=query)
response = llm.invoke(messages)

print(response.content)
```

```python
def get_sources(docs):
    pages = []
    for doc in docs:
        page = doc.metadata.get("page", "N/A")
        if page not in pages:
            pages.append(page)
    return pages

sources = get_sources(docs)
print("Sources / Pages:", sources)
```

```python
print("Answer:\n", response.content)
print("\nSources / Pages:", sources)
```

```python
def evaluate_rag(question, retriever, llm):
    docs = retriever.invoke(question)
    context = format_docs(docs)
    messages = prompt.format_messages(context=context, question=question)
    answer = llm.invoke(messages).content

    return {
        "question": question,
        "answer": answer,
        "sources": get_sources(docs),
        "context": context
    }

result = evaluate_rag("What topics are discussed in the document?", retriever, llm)
print(result["answer"])
print(result["sources"])
```
